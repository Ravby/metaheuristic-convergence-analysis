# Rating interval bands for CEC 2024 problems with additional algorithms

#### Lets-Plot setup

Because this notebook lives inside the Gradle module, the module's classpath (including the **Lets-Plot** version declared in `build.gradle.kts`) is already available to the kernel. So instead of the `%use lets-plot` line magic, Lets-Plot is set up directly from that classpath.

The line magic resolves its own Lets-Plot version from a library descriptor, which is not pinned to the project's version and keeps moving with new releases. Once the descriptor gets ahead of the version on the module classpath, loading fails with `NoSuchMethodError: LetsPlot.setOutputsDescription`. Setting the library up by hand keeps the notebook tied to exactly one version — the one in `build.gradle.kts`.

The imports below also bring in `ggtb()`, which enables zoom/pan interactivity on the plot.

```kotlin
import org.jetbrains.letsPlot.*
import org.jetbrains.letsPlot.export.*
import org.jetbrains.letsPlot.geom.*
import org.jetbrains.letsPlot.interact.*
import org.jetbrains.letsPlot.label.*
import org.jetbrains.letsPlot.scale.*
import org.jetbrains.letsPlot.themes.*

// Keep in sync with build.gradle.kts: lets-plot-kotlin 4.15.0 pairs with the 4.11.0 core/JS bundle.
val letsPlotFrontend = LetsPlot.setupNotebook("4.11.0", null) { DISPLAY(HTML(it)) }
LetsPlot.apiVersion = "4.15.0"
DISPLAY(HTML(letsPlotFrontend.getConfigureHtml()))
```

#### Results directory path

Define the path to the results directory for CEC 2024 experiments.

```kotlin
val resultsDirectory = "${System.getProperty("user.dir")}../../../../../../../experimental_results/CEC2024/With additional algorithms"
```

#### Load results

Load algorithm performance results from the specified directory.

```kotlin
val algorithms = si.um.feri.lpm.loadAlgorithmResults(si.um.feri.lpm.BenchmarkId.CEC2024, resultsDirectory)
```

#### Plot generation

Generate a plot for the loaded algorithms with specified x-axis breaks.

```kotlin
var plot = si.um.feri.lpm.generatePlot(
    algorithms = algorithms,
    xAxisMode = si.um.feri.lpm.XAxisMode.BUDGET_PERCENT,
    cutpointStep = 10
)
```

#### Display the plot

Add zoom and pan interactivity to the plot and display the plot.

```kotlin
plot += ggtb()
plot.show()
```

#### Save the plot

Save the plot as an SVG file in the specified directory.

```kotlin
ggsave(plot, path = resultsDirectory, filename = "CEC2024_100CP_rating_interval_bands_additional_algorithms.svg")
ggsave(plot, path = resultsDirectory, filename = "CEC2024_100CP_rating_interval_bands_additional_algorithms.png")
```
